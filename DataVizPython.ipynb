# Data Visualization in Python

- Matplotlib
- Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
titanic
```

```python
titanic.info()
```

```python
titanic.isnull().sum()
```

## Matplotlib Demo

```python
# survival in matplotlib


survival_counts = titanic["survived"].value_counts()
plt.bar(survival_counts.index,survival_counts.values,color=["pink","skyblue"])
plt.xticks([0,1],["Not Survived","Survived"])
plt.title("Survival Count")
plt.ylabel("Number of passsengers")
plt.xlabel("Survived Yes or No")
plt.show()
```

```python
sns.set_style("whitegrid")
```

```python
# distribution of pclass in matplotlib


pclass_counts = titanic["pclass"].value_counts()
plt.bar(pclass_counts.index,pclass_counts.values,color=["salmon","green","purple"])
plt.xticks([3,1,2],["Third","First","Second"])
plt.title("Passenger Class")
plt.ylabel("Number of passsengers")
plt.xlabel("Class of the passenger")
plt.show()
```

```python
# Survival by Gender

gender_survival = titanic.groupby(["sex","survived"]).size().unstack()
gender_survival
```

```python
gender_survival.plot(kind = "bar",stacked = False, color = ["skyblue","salmon"])
plt.show()
```

```python
# Age distribution

plt.hist(titanic["age"].dropna(),color="red")
plt.show()
```

## Seaborn Demo

```python
# Survival rate by Gender

sns.barplot(x = "sex", y="survived",data=titanic,palette="pastel")
plt.show()
```

```python
# class vs survived

sns.barplot(x = "class", y="survived",data=titanic,palette="muted")
plt.show()
```

```python
sns.histplot(data = titanic, x="age",hue="survived",multiple="stack",palette="Set1")
plt.title("Age Distribution")
plt.show()
```

```python
# boxplot

sns.boxplot(x="survived",y="fare",data=titanic,hue="sex",palette="Set2")
plt.title("Fare distribution by Survival")
plt.show()
```

```python
#scatterplot

sns.scatterplot(x="fare",y="age",data = titanic,color="blue",hue="sex")
```

```python
# jointplot

sns.jointplot(x="fare",y="age",data = titanic,color="purple")
plt.show()
```

```python
sns.histplot(titanic["fare"],color="red")
plt.show()
```

```python
sns.boxplot(x="class",y="age",data=titanic,hue="class",palette="rainbow")
plt.show()
```

```python
sns.countplot(x="sex",data=titanic,palette="muted",hue="sex")
plt.show()
```

```python
sns.countplot(x="class",data=titanic,palette="Set3",hue="sex")
plt.show()
```

```python
sub = titanic[["age","fare"]]
sub
```

```python
sub.corr()
```

```python
sns.heatmap(sub.corr(),annot=True)
plt.show()
```

# Using Gen AI coding

### Plotly Demo

```python
import plotly.express as px
```

```python
# Survival count

fig = px.pie(
    titanic,
    names='survived',
    title='Survivors vs Non-Survivors'
)
fig.show()
```

```python
# passenger class count

fig = px.pie(
    titanic,
    names='pclass',
    title='Passenger Class'
)
fig.show()
```

```python
# Histogram in plotly
fig = px.histogram(
    titanic,
    x='class',
    color='class',
    title='Passenger Distribution by Class'
)
fig.show()
```

```python
# survival by Gender
fig = px.histogram(
    titanic,
    x='sex',
    color='survived',
    barmode='group',
    title='Survival by Gender'
)
fig.show()
```

```python
# Fare distribution by Class

fig = px.box(
    titanic,
    x='class',
    y='fare',
    color='class',
    title='Fare Distribution by Passenger Class'
)
fig.show()
```
