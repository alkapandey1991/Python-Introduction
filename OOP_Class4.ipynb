## Object Oriented Programming in Python

```python
# A class can have functions and attributes:

class employee:
    def work(self):
        print("This employee has been added to the organization")
    
```

```python
# NOTE: If you want to access a function within a class, it can only be accessed by an object of the same class
```

```python
# Calling function defined within the class:

# step 1: Create an object of the class
# step 2: Using the object created, call the function

employee1 = employee() #To create an object of the class, use the class name

employee1.work()
```

# Example 2: With constructor


##### Note: "Self" keyword is used to point to the current object 

```python
# A class can have functions and attributes:

class employee:
    def __init__(self): #A constructor is a function which gets executed by default automatically when you create a class
        self.name = input("Enter the name of the new employee:")
        print("Welcome to the new organization",self.name)
    def salarydef(self):
        self.salary = input("Enter the salary:")
        print("Congratulations! Your salary is:",self.salary)
        
    
```

```python
e1 = employee()
```

```python
e1.salarydef()
```

```python
e2 = employee()
```

```python
e2.salarydef()
```

```python
e1.name
```

```python
e2.name
```

```python
e1.__dict__
```

## Inheritance:

```python
class Animal(): #parent class
    def speak(self):
        print("Animal is speaking...")

class Dog(Animal): #child class
    def bark(self):
        print("The Dog Barks: Woof woof woof....!!!!")
```

```python
a1 = Animal()
a1.speak()
```

```python
dog1 = Dog()
dog1.bark()
```

```python
dog1.speak()
```

### Multiple Inheritance

```python
class Father():
    def skills(self):
        print("Gardening and Plumbing")

class Mother():
    def hobbies(self):
        print("Painting and Coding")

class Child(Father,Mother):
    def talent(self):
        print("Singing and Dancing")
```

```python
c1 = Child()
```

```python
c1.talent()
```

```python
c1.hobbies()
```

```python
c1.skills()
```
