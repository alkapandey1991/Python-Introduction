## Generator functions and Regular Expression


### Class 3 Python

```python
# Traditional function:

def create_gene(n):
    mylist = []
    for i in range(5):
        mylist.append(i*2)
    return mylist
```

```python
# call function

saved = create_gene(5)
print(saved)
```

```python
# Using generator function with yield instead of return

def create_gener(n):
    for i in range(n):
        yield i*2
```

```python
# function call and usage

saved2 = create_gener(5)
print(saved2)
```

# Regular Expression:

```python
import re
```

```python
text = "I am doing well in the class and I want to go home"
match = re.search("well",text)
print(match.group())
```

```python
match.start() # return the start index location where the match is found
```

```python
match.end() # return the end index location where the match is found
```

```python
text2 = "I am doing well in the class and I want to go home wellington wellwisher wellbeing welljfsahgdhagfdaghdfashgdsasgh"
match = re.search("we*",text)
print(match.group())
```

```python
f = re.findall(r"well[A-Za-z]{0,10}",text2)
print(f)
```

```python
# re.search - Searches a pattern anywhere in a text

# re.match - returns the first match beginning of a string, and you can also retrieve the start index match and end index match

# re.findall - find all matches in a text

# re.sub - replace a pattern
```

### Search

```python
text3 = "Python is not so great and so many things are different in Pythono. I don't like Python"
pattern = "Python[a-z]"

result = re.search(pattern,text3)

print(result)
```

```python
print(result.start())
print(result.end())
```

```python
text3 = "Python is not so Gr8 and so many things are different in 9 Python. I don't like Python 100"
pattern = "\d"

result = re.search(pattern,text3)

print(result)
```

### Findall

```python
# re.findall

text3 = "Python is not so Gr8 and so many things are different in 95 Python. I don't like Python 100"
pattern = "\d{1,5}"

result = re.findall(pattern,text3)

print(result)
```

```python
# re.findall

text3 = "My phone number is 9797654312 and my friends numbers is 9791444555 767"
pattern = "\d{10}"

result = re.findall(pattern,text3)

print(result)
```

### Match

```python
# re.match

text3 = "Python is not so Gr8 and so many things are different in 95 Python. I don't like Python 100"
pattern = "Python"

result = re.match(pattern,text3)

print(result)
```

### Substitute or Replace based on a pattern

```python
# re.sub

#example to remove/replace digits with empty string
text = "Order123 Amount456"
result = re.sub(r"\d+", "", text)

print(result)
```
