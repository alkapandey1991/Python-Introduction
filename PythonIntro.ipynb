### Conditional Statements

```python
# if statement:

a = int(input("Enter number 1:"))
b = int(input("Enter number 2:"))

if a<b:
    print("First number is less than the second")
    print("This is another line in the code block of if")
    print("******************")
else:
    print("Second number is larger than the first")
    
```

```python
a = 8
b = 9
c = 10

if a < b:
    print("a is less than b")
if a < c:
    print("a is less than c")
```

```python
#elif statement:
a = 1
b = 5
c = 14
if a<b and a<c:
    print("a is the smallest")
elif b<a and b<c:
    print("b is the smallest")
else:
    print("c is the smallest")
```

## Loops:

```python
fruits = ["Apple","Banana","Orange","Mango"]

for f in fruits:
    print(f,"is available")
```

```python
list1 = [3,4,7,1,9]
for num in list1:
    print("The number is {} and the square of the number is {}".format(num,num**2))
```

```python
for i in range(1,15):
    n = int(input("Enter the number"))
    if n == 0:
        print("You have chosen to exit the process!")
        break
    else:
        continue
```

```python
for i in range(1,11):
    if i==5:
        continue
        print("**")
    else:
        print(i)
```

```python
# While:

j = 1
while(j<10):
    print(j,"is less than 10")
    j = j+1
```
