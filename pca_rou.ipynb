# PCA (Principal Component Analysis on Iris Dataset)

### Note:

1. PCA is only applied on "X" variables
2. Number of X variables = No of PCs generated
3. You need to select the optimum number of PC to represent your data
4. All X variables should be scaled

## Alka Pandey

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("iris.csv")
data
```

```python
data.pop("Id")
```

```python
data
```

```python
y = data.pop("Species")
y
```

```python
data.info()
```

```python
data.describe()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X = data.copy()
x = scaler.fit_transform(X)
```

```python
x
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
```

```python
pca.fit(x)
```

```python
pca.components_
```

```python
pca.explained_variance_ratio_
```

```python
# Variance explained by each principal component

plt.bar(range(1,5),pca.explained_variance_ratio_)

plt.title("Variance Explained by Principal Components")
plt.ylabel("Variance Explained")
plt.xlabel("Principal Components")

plt.show()
```

```python
var_cumu = np.cumsum(pca.explained_variance_ratio_)
var_cumu
```

```python
# scree plot

plt.plot(range(1,5),var_cumu)
plt.show()
```

```python
pc2 = PCA(n_components=2)
```

```python
newdata = pc2.fit_transform(x)
newdata
```

```python
df = pd.DataFrame(newdata,columns=["PC1","PC2"])
df
```

```python
df_final = pd.concat([df,y],axis=1)
df_final
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
X = df_final[["PC1","PC2"]]
y = df_final["Species"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20)
```

```python
log = LogisticRegression()
log.fit(X_train,y_train)
y_pred = log.predict(X_test)
```
