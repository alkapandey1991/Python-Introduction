### Concat

```python
df1 = pd.DataFrame({"ID":[1,2,3],"Name":["Alice","Bob","Charlie"],"Age":[25,26,30]})
df1
```

```python
df2 = pd.DataFrame({"ID":[4,5,6,7],"Name":["David","Eve","Frank","Gilbert"],"Age":[31,35,32,39]})
df2
```

```python
# STACK DATAFRAMES VERTICALLY WITH CONCAT

stackedvertical = pd.concat([df1,df2],ignore_index=True,axis=0) #axis = 0 indicates to python that stacking should be row wise
stackedvertical
```

```python
df3 = pd.DataFrame({"ID":[1,2,3,4],"Name":["David","Eve","Frank","Gilbert"]})
df3
```

```python
df4 = pd.DataFrame({"Age":[31,35,32,39],"Salary":["90k","100k","120k","89k"]})
df4
```

```python
stackedhori = pd.concat([df3,df4],axis=1) #axis = 1 indicates to python that stacking should be column wise
stackedhori
```

### Merge

```python
df5 = pd.DataFrame({"ID":[1,2,3,4],"Name":["Alice","Bob","Charlie","David"],"Salary":["25k","26k","30k","40k"]})
df5
```

```python
df6 = pd.DataFrame({"ID":[3,4,5,6],"Name":["Charlie","David","Eve","Frank"],"Age":[40,45,42,49]})
df6
```

```python
# inner join - common for both dataframes:

innerj = pd.merge(df5,df6,how="inner",on="ID")
innerj
```

```python
innerjclean = innerj.drop(["Name_y"],axis=1)
innerjclean
```

```python
innerjclean.columns = ["ID","Name","Salary","Age"]
innerjclean
```

```python
# left join - keep all values from left and match with right:

leftj = pd.merge(df5,df6,how="left",on="ID")
leftj
```

```python
# right join - keep all values from right and match with left:

rj = pd.merge(df5,df6,how="right",on="ID")
rj
```

```python
# right join - both left and right with corresponding match

oj = pd.merge(df5,df6,how="outer",on="ID")
oj
```

```python
oj.dropna()
```

```python
import pandas as pd

data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East', 'West', 'West'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile', 'Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 30000, 45000, 25000, 40000, 22000, 35000, 28000],
    'Quantity': [10, 15, 8, 12, 9, 11, 7, 14]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Sum of sales by Region:

pd.pivot_table(
    df,
    index='Region', #The column you want to see in rows
    values='Sales', #The value you want to aggregate
    aggfunc='sum' #The function to aggregate on
)
```

```python
# Sales by Region and product

pd.pivot_table(
    df,
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum'
)
```

```python
# Sales by Region and product

pd.pivot_table(
    df,
    index=['Region','Product'],
    values='Sales',
    aggfunc='sum'
)
```

```python
# Multiple Aggregations:

pd.pivot_table(
    df,
    index='Region',
    values='Sales',
    aggfunc=['sum', 'mean', 'max']
)
```

```python
# Multiple values to aggregate:

pd.pivot_table(
    df,
    index='Region',
    values=['Sales', 'Quantity'],
    aggfunc='sum'
)
```
