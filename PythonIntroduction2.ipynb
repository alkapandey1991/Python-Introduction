```python
# map : applies a function to all items in a sequence
```

```python
numb = [2,3,4,5,6]
print(numb)
```

```python
list1 = list(map(lambda x:x**3,numb))
list1
```

```python
def myfunction(num1):
    s = num1+num1*2
    return s
```

```python
list(map(myfunction,numb))
```

```python
# filter

list2 = list(filter(lambda x:x%2==0,numb))
list2
```

```python
# reduce:

from functools import reduce

reduce(lambda x,y:x+y,numb)
```

## Exception Handling using try-except-else-finally

```python
try:
    num1 = int(input("Enter number 1:"))
    num2 = int(input("Enter number 2:"))
    res = num1/num2
    print("The division of numbers is:",res)
except ZeroDivisionError:
    print("Error Encountered!!! You cannot divide by zero, Please enter another number...")
except ValueError:
    print("Not a valid error...!!!")
else:
    print("The division was performed successfully!!!")
finally:
    # Always executes
    print("**Program Completed**")
```
